## Exploratory Visualizations 

```python
numeric_features = df.select_dtypes(include=["int64", "float64"])
corr_matrix = numeric_features.corr()
sns.heatmap(corr_matrix, annot=True)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(
    df,
    hue="dlq_2yrs",
    diag_kind="hist",
    plot_kws={"alpha": 0.3, "s": 15},
)

plt.suptitle("Scatterplot Matrix of Credit Risk Features", y=1.02)
plt.show()
```

```python
# target variable distribution
df["dlq_2yrs"].hist(
    figsize=(10, 8),
    bins=30,
    edgecolor="black"
)
plt.suptitle("Target Variable Distribution", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
features = df.drop(columns=["dlq_2yrs"]).columns

df[features].hist(
    figsize=(16, 12),
    bins=30,
    edgecolor="black"
)

plt.suptitle("Distribution of Numerical Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 8))

sns.boxplot(data=df[features], orient="h")

plt.title("Boxplots of Numerical Features")
plt.xlabel("Value")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(4, 3, figsize=(16, 14))
axes = axes.flatten()

for i, col in enumerate(df.columns):
    sns.kdeplot(
        data=df,
        x=col,
        fill=True,
        ax=axes[i]
    )
    axes[i].set_title(f"Density Plot: {col}")

# Remove unused subplot
for j in range(len(df.columns), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
features = df.drop(columns=["dlq_2yrs"]).columns

fig, axes = plt.subplots(4, 3, figsize=(16, 14))
axes = axes.flatten()

for i, col in enumerate(features):
    sns.kdeplot(
        data=df,
        x=col,
        hue="dlq_2yrs",
        fill=True,
        common_norm=False,
        alpha=0.3,
        ax=axes[i]
    )
    axes[i].set_title(f"{col} by Delinquency")

for j in range(len(features), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
